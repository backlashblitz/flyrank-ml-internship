## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Evaluation Metric: PR-AUC (Precision-Recall Area Under Curve) & Precision@K (e.g., Precision@50)Why PR-AUC over ROC-AUC / Accuracy:Class Imbalance: Declining pages represent a minority subset of all published site content. High accuracy can be achieved trivially by predicting "not declining".Operational Efficiency: Editorial teams have limited weekly capacity and can only audit the top $K$ flagged pages. High Precision@K ensures that when an editor reviews the top 50 flagged URLs, the majority are genuinely decaying pages that benefit from updates.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Limitations of Simple Rules: A heuristic rule like "flag any page un-updated for 180+ days" generates high false positives. Many evergreen pages remain highly ranked without frequent updates, while fast-moving topics decay within weeks. Rules cannot weigh non-linear interactions between CTR drops, seasonal shifts, and positional drifts.

Why ML Wins: A trained machine learning model combines temporal trends, keyword positioning, interaction history, and content age into a unified probability score. It adapts dynamically to subtle multi-feature signals, ranking decaying pages far more accurately than static hard-coded thresholds.
